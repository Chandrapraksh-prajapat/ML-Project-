# Phase 6: Outlier Detection & Treatment

Features

Analyze :

Rating,
Reviews,
Installs,
Price

Tasks :

1. Visualize outliers using boxplots.
2. Detect outliers using the IQR method.
3. Detect outliers using Z-score.
4. Remove outliers.
5. Apply outlier capping.
6. Compare both approaches.

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
cols=['Rating','Reviews','Installs','Price']
```

```python
# Visualize outliers using boxplots.

for col in cols:
  plt.figure(figsize=(6,3) )
  sns.boxplot(x=df[col])
  plt.title(col)
  plt.show()
```

```python
# Detect outliers using the IQR method

for col in cols:
  Q1=df[col].quantile(0.25)
  Q3=df[col].quantile(0.75)
  IQR=Q3-Q1

  lower_bound=Q1-1.5*IQR
  upper_bound=Q3+1.5*IQR

  outliers=df[(df[col]<lower_bound) | (df[col]>upper_bound)]
  print(col,":",len(outliers))

```

```python
# # Detect outliers using Z-score.
```

```python
# Remove outliers

df_removed=df.copy()

for col in cols:

  Q1=df[col].quantile(0.25)
  Q3=df[col].quantile(0.75)

  IQR=Q3-Q1

  lower_bound=Q1-1.5*IQR
  upper_bound=Q3+1.5*IQR

  df_removed = df_removed [(df_removed[col] >= lower_bound) & (df_removed[col] <= upper_bound)]

  print(col,":",len(df_removed))
```

```python
print(df.shape)
print(df_removed.shape)
```

```python
# Apply outlier capping.

df_capping=df.copy()

for col in cols:

  Q1=df[col].quantile(0.25)
  Q3=df[col].quantile(0.75)
  IQR=Q3-Q1

  lower_bound=Q1-1.5*IQR
  upper_bound=Q3+1.5*IQR

  df_capping[col]=df_capping[col].clip(lower_bound,upper_bound)

```

```python
# for col in cols:
#   plt.figure(figsize=(6,3) )
#   sns.boxplot(x=df_capping[col])
#   plt.title(col)
#   plt.show()
```

```python
#  compare both approches

print("Original dataset : ",df.shape )
print("After removing outliers :",df_removed.shape)
print("After outlier capping : ", df_capping.shape)
```


#Phase 11: Feature Creation

Create the following features.

Feature 1
: Reviews_Per_Install

Formula :

 Reviews / Installs

Feature 2
: Paid_App_Flag

Paid = 1
Free = 0

Feature 3
: App_Age

Formula:

Current Year - Update Year

Feature 4
: Popularity_Score

Formula:

Rating * Reviews

Feature 5
: Revenue_Potential

Formula:

Price * Installs

Feature 6
: Category_Average_Rating

```python
# Feature 1 : Reviews_Per_Install
# df["Reviews_Per_Install"]=df["Reviews"]/df["Installs"]

import numpy as np

df["Reviews_Per_Install"] = np.where(df["Installs"] == 0, 0, df["Reviews"] / df["Installs"])
```

```python
df[["Reviews","Installs","Reviews_Per_Install"]].head(2)
```

```python
# Feature 2 : Paid_App_Flag
df["Paid_App_Flag"]=df['Type'].astype(int)
```

```python
# Feature 3 : App_Age
#  formula : current year - update year

import datetime

current_year = datetime.datetime.now().year

df["App_Age"] = current_year - df["Updated_Year"]
```

```python
df[["Updated_Year","App_Age"]].head(2)
```

```python
# Feature 4: Popularity_Score

#  formula :  popularity_score= Rating* Reviews

df["Popularity_Score"]=df["Rating"]*df["Reviews"]
```

```python
df[["Rating","Reviews","Popularity_Score"]].head(2)
```

```python
# Feature 5: Revenue_Potential

# Formula : Revenue_Potential= Price * Installs

df["Revenue_Potential"]=df["Price"]*df["Installs"]
```

```python
df[["Price","Installs","Revenue_Potential"]].head(2)
```

```python
#  Feature 6 : category_Average_Rating

# calculate the average rating for each category and assign it to every app
```

```python
df_features["Category_Average_Rating"] =(
        df_features.groupby("Category")["Rating"].transform("mean"))
```
